```python
import matplotlib.pyplot as plt

categories = ['Bar 1', 'Bar 2']
values = [400000, 500000]

plt.bar(categories, values, color=['blue', 'orange'])
plt.title('Chart with Two Bars')
plt.show()

plt.bar(['Bar 1', 'Bar 2'], [250, 300], color=['blue', 'orange'])
plt.title('Chart with Two Bars')
plt.show()
```
